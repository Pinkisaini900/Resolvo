```python
import os, io, getpass
import gradio as gr
from google import genai
from google.genai import types

# ---------- API key ----------
API_KEY = getpass.getpass("Apni Gemini API key paste karein aur Enter dabayein: ").strip()
client = genai.Client(api_key=API_KEY)

# ---------- Model apne aap chunna (jo chal jaye wahi) ----------
def pick_model():
    cands = []
    for m in client.models.list():
        n = m.name.replace("models/", "")
        acts = getattr(m, "supported_actions", None) or []
        if acts and "generateContent" not in acts:
            continue
        if any(x in n for x in ["image", "tts", "live", "audio", "embedding",
                                "omni", "robotics", "computer", "veo", "imagen"]):
            continue
        if "flash" in n or "lite" in n:
            cands.append(n)
    cands.sort(reverse=True)
    print("Try karne wale models:", cands[:12])

    for n in cands[:12]:
        try:
            client.models.generate_content(model=n, contents="hi")
            print("✅ Chal gaya:", n)
            return n
        except Exception as e:
            print("❌ Skip:", n, "->", str(e)[:70])
    raise RuntimeError("Kisi bhi model par free quota nahi mila.")

MODEL = pick_model()
print("✅ Model chuna gaya:", MODEL)

SYSTEM = """Aap ek expert plant pathologist aur krishi salahkar ho. User ek patte ki photo bhejega.
Hinglish (Hindi Roman script me, simple bhasha) me jawab do.

Photo ke liye is format me jawab do:
🌱 **Paudhe ka naam:** (common naam + scientific naam)
🏷️ **Type:** (fasal / sabzi / phal / sajawati / jungli / anya)
🩺 **Sehat:** (swasth / bimar / pakka nahi)
🦠 **Bimari ya samasya:** (naam, ya "koi nahi")
🔍 **Lakshan:** (photo me kya dikh raha hai)
❓ **Karan:** (fungus, bacteria, virus, keet, pani/khaad ki kami, etc.)
💊 **Rasayanik upchar:** (dawai ke naam)
🌿 **Jaivik upchar:** (neem tel, etc.)
🛡️ **Rokthaam:**
📊 **Bharosa:** (kam / madhyam / zyada, aur kyun)

Niyam:
- Agar photo me patta nahi hai, to saaf bolo ki ye patte ki photo nahi lag rahi.
- Agar pakka nahi ho, to bolo, aur 2-3 sambhavit options batao. Andaza lagakar jhooth mat bolo.
- Dawai ke saath hamesha bolo ki matra ke liye local krishi visheshagya/KVK se poochein.
- Follow-up sawalon ka seedha aur chhota jawab do."""

# ---------- Helper functions ----------
def image_bytes(image):
    img = image.convert("RGB")
    img.thumbnail((1568, 1568))
    buf = io.BytesIO()
    img.save(buf, format="JPEG", quality=90)
    return buf.getvalue()

def ask(history):
    contents = []
    for m in history:
        parts = []
        for p in m["parts"]:
            if p["type"] == "image":
                parts.append(types.Part.from_bytes(data=p["data"], mime_type="image/jpeg"))
            else:
                parts.append(types.Part.from_text(text=p["text"]))
        contents.append(types.Content(role=m["role"], parts=parts))

    resp = client.models.generate_content(
        model=MODEL,
        contents=contents,
        config=types.GenerateContentConfig(system_instruction=SYSTEM),
    )
    return resp.text or "Koi jawab nahi mila. Dobara try karein."

def render(history):
    out = []
    for m in history:
        text = " ".join(p.get("text", "") for p in m["parts"] if p["type"] == "text")
        has_img = any(p["type"] == "image" for p in m["parts"])
        if m["role"] == "user":
            out.append("### 📷 Aapne photo bheji" if has_img else f"### 🙋 Aap\n{text}")
        else:
            out.append(f"### 🤖 AI\n{text}")
    return "\n\n---\n\n".join(out)

def analyze(image):
    if image is None:
        return "📷 Pehle patte ki photo upload karein.", []
    try:
        history = [{
            "role": "user",
            "parts": [
                {"type": "image", "data": image_bytes(image)},
                {"type": "text", "text": "Is patte ko pehchaan kar poora analysis do."},
            ],
        }]
        reply = ask(history)
        history.append({"role": "model", "parts": [{"type": "text", "text": reply}]})
        return render(history), history
    except Exception as e:
        return f"❌ Error: {e}", []

def chat(message, history):
    if not history:
        return "📷 Pehle photo upload karke Analyze dabayein.", history, ""
    if not message or not message.strip():
        return render(history), history, ""
    try:
        new_hist = history + [{"role": "user", "parts": [{"type": "text", "text": message}]}]
        reply = ask(new_hist)
        new_hist.append({"role": "model", "parts": [{"type": "text", "text": reply}]})
        return render(new_hist), new_hist, ""
    except Exception as e:
        return f"❌ Error: {e}", history, message

def clear_all():
    return None, "Result yahan dikhega.", [], ""

# ---------- Interface ----------
with gr.Blocks(title="Leaf Doctor AI") as demo:
    gr.HTML("""
    <div style="text-align:center;padding:10px 0;">
      <h1 style="margin:0;color:#166534;">🌿 Leaf Doctor AI</h1>
      <p style="margin:4px 0 0 0;color:#4b5563;">
        Koi bhi patte ki photo daalein. Paudhe ka naam, bimari aur ilaaj paayein, aur chat se sawaal poochein.
      </p>
    </div>""")

    state = gr.State([])

    with gr.Row():
        with gr.Column(scale=1):
            img = gr.Image(type="pil", label="📷 Patte ki photo")
            analyze_btn = gr.Button("🔍 Analyze karein", variant="primary")
            clear_btn = gr.Button("🗑️ Clear")
            gr.Markdown("**Tips:** ek patta • saaf roshni • bimar hissa close-up me • dhundhli photo na ho")

        with gr.Column(scale=2):
            output = gr.Markdown("Result yahan dikhega.")
            msg = gr.Textbox(label="💬 Follow-up sawaal poochein",
                             placeholder="Jaise: Spray kab karein? Kitni matra? Ye dusre paudhon me failega kya?")
            send_btn = gr.Button("Bhejein")

    analyze_btn.click(analyze, inputs=img, outputs=[output, state])
    send_btn.click(chat, inputs=[msg, state], outputs=[output, state, msg])
    msg.submit(chat, inputs=[msg, state], outputs=[output, state, msg])
    clear_btn.click(clear_all, inputs=None, outputs=[img, output, state, msg])

    gr.Markdown("⚠️ Ye AI sirf margdarshan ke liye hai. Dawai se pehle krishi visheshagya/KVK se salah lein.")

demo.launch(share=True, debug=True)
```
